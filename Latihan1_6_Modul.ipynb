```python
#Latihan 1
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name # Nama Hero
        self.hp = hp # Nyawa (Health Point)
        self.attack_power = attack_power # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method
hero1.info()
hero2.info()

# Mengubah nilai hp hero1
hero1.hp = 500
print(f"New HP for hero1: {hero1.hp}")

# Latihan 2

# Tambah kode Output di akhir program
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas
```

```python
# Latihan 3
class Hero:
  def __init__(self, name, hp, attack_power):
    self.name = name
    self.hp = hp
    self.attack_power = attack_power

  def info(self):
    print(f"{self.name} [Hero] | HP: {self.hp}")

  def serang(self, lawan):
    print(f"{self.name} menyerang {lawan.name} dengan {self.attack_power} damage!")
    lawan.diserang(self.attack_power)

  def diserang(self, damage):
    self.hp -= damage
    print(f"{self.name} diserang dan menerima {damage} damage! HP tersisa: {self.hp}")
    if self.hp <= 0:
      print(f"{self.name} telah dikalahkan!")

# Class Mage adalah anak dari class Hero
class Mage(Hero):
  def __init__(self, name, hp, attack_power, mana):
    # Memanggil constructor milik Parent (Hero)
    super().__init__(name, hp, attack_power)
    self.mana = mana

  def info(self):
    print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

  # Mage punya skill khusus
  def skill_fireball(self, lawan):
    if self.mana >= 20:
      print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
      self.mana -= 20
      lawan.diserang(self.attack_power * 2) # Damage 2x lipat
    else:
      print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)       # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage
```

```python
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di
        # dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar
        # aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)
# hero1.__hp = 9999 # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp) # ERROR (Tidak bisa dibaca langsung)
hero1.set_hp(-50) # Coba set negatif
print(hero1.get_hp()) # Output: 0 (Karena dicegat oleh logika Setter)
print(f"Mencoba akses paksa: {hero1._Hero__hp}")
```

```python
#Uji Validaasi Hapus logika if dan elif di dalam method set_hp, sehingga isinya hanya self.__hp = nilai_baru.
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di
        # dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        # Logika if dan elif dihapus sesuai permintaan
        self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar
        # aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)
# hero1.__hp = 9999 # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp) # ERROR (Tidak bisa dibaca langsung)
hero1.set_hp(-100) # Mengubah HP menjadi negatif
print(hero1.get_hp()) # Output setelah perubahan
print(f"Mencoba akses paksa: {hero1._Hero__hp}")
```

```python
from abc import ABC, abstractmethod
# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass
    @abstractmethod
    def info(self):
        pass
# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama
    # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")
    def info(self):
        print(f"Saya adalah Hero: {self.nama}")
class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis
    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")
    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")
# -- Uji Coba --
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()
```

**Penjelasan Error `Can't instantiate abstract class Hero with abstract method...?`**

Error *abstract class* pada dasarnya terjadi karena kelas turunan (seperti *Hero*) belum menepati "kontrak" atau aturan wajib yang ditetapkan oleh kelas induknya. Ibarat memproduksi mainan tanpa memasang tombol utama, Python akan menolak membuat objek tersebut dan menganggap kodenya belum selesai jika ada metode wajib (seperti *serang* atau *info*) yang belum kamu buat. Oleh karena itu, agar karakter bisa berhasil dibuat tanpa eror, seluruh metode wajib dari kelas induk harus ditulis dan diimplementasikan secara utuh terlebih dahulu.

**Pertanyaan: Mengapa class `GameUnit` dilarang untuk dibuat menjadi objek?**

Jika Anda mengaktifkan baris `unit = GameUnit()`, Anda akan mendapatkan `TypeError: Can't instantiate abstract class GameUnit with abstract methods info, serang`.

Alasannya adalah:

*   **`GameUnit` adalah Kelas Abstrak (Abstract Class):** Kita telah mendeklarasikan `GameUnit` sebagai kelas abstrak dengan mewarisi dari `ABC` (Abstract Base Classes) dan menandai method `serang` dan `info` dengan `@abstractmethod`. Kelas abstrak adalah "cetak biru" atau "kontrak" yang belum lengkap.
*   **Method Abstrak Belum Diimplementasikan:** Sebuah kelas abstrak memiliki satu atau lebih method abstrak yang belum memiliki implementasi (hanya dideklarasikan dengan `pass`). Python mencegah pembuatan objek dari kelas abstrak karena ia tahu bahwa kelas tersebut belum menyediakan semua fungsionalitas yang dijanjikan oleh method abstraknya.

**Apa gunanya ada class `GameUnit` jika tidak bisa dibuat menjadi objek nyata?**

Kelas `GameUnit` sangat berguna meskipun tidak bisa dibuat menjadi objek langsung, yaitu:

1.  **Definisi Antarmuka (Interface / Kontrak):** `GameUnit` berfungsi sebagai kontrak atau antarmuka. Ini mendefinisikan bahwa **setiap kelas konkret** (seperti `Hero` dan `Monster`) yang mewarisi dari `GameUnit` **wajib** memiliki method `serang` dan `info`. Ini memastikan semua "unit game" berperilaku seragam dalam hal kemampuan dasar ini.
2.  **Meningkatkan Konsistensi Kode:** Dengan memaksa implementasi method tertentu, `GameUnit` memastikan konsistensi di antara semua kelas turunannya. Anda tahu pasti bahwa objek `Hero` akan memiliki `serang()` dan `info()`, begitu pula objek `Monster`.
3.  **Polimorfisme:** Ini memungkinkan kita untuk menulis kode yang bekerja dengan objek dari berbagai kelas turunan `GameUnit` secara generik. Misalnya, Anda bisa memiliki daftar `game_units = [hero1, monster1, hero2]` dan mengiterasi daftar tersebut, memanggil `unit.serang(target)` pada setiap unit tanpa perlu tahu apakah itu `Hero` atau `Monster`. Python akan memanggil implementasi `serang` yang benar untuk objek tersebut.
4.  **Desain yang Jelas:** Membantu dalam merancang hierarki kelas yang jelas dan terstruktur, di mana aturan dasar didefinisikan di tingkat abstrak dan implementasi spesifik diserahkan kepada kelas konkret.

```python
# Latihan 6

# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")

# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")

# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

# Child Class 4
class Healer(Hero):
    def serang(self):
        print(f"{self.nama} (Healer) tidak menyerang, tapi menyembuhkan teman!")


# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]

print("--- PERANG DIMULAI ---")
# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()


# --- UJI COBA: Polimorfisme Berhasil ---
# PERBAIKAN: Selalu beri argumen 'nama' saat instansiasi objek
pasukan = [Hero("Layla"), Archer("Miya"), Mage("Eudora")]

print("\n--- UJI COBA 2 ---")
# Pemanggil tidak perlu tahu detail jurus masing-masing, cukup panggil 'serang()'
for unit in pasukan:
    unit.serang()
```
